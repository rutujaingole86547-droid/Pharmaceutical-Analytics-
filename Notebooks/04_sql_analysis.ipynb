```python
import pandas as pd

customers = pd.read_csv("../Data/Cleaned/customers_clean.csv")
products = pd.read_csv("../Data/Cleaned/products_clean.csv")
suppliers = pd.read_csv("../Data/Cleaned/suppliers_clean.csv")
inventory = pd.read_csv("../Data/Cleaned/inventory_clean.csv")
sales = pd.read_csv("../Data/Cleaned/sales_transactions_clean.csv")
```

```python
#supplier performance summary
query = """
SELECT

sp.supplier_name,

sp.supplier_rating,

sp.delivery_days,

COUNT(s.order_id) AS Total_Orders,

ROUND(SUM(s.revenue),2) AS Revenue,

ROUND(SUM(s.profit),2) AS Profit

FROM suppliers sp

LEFT JOIN sales_transactions s

ON sp.supplier_id = s.supplier_id

GROUP BY
sp.supplier_id,
sp.supplier_name,
sp.supplier_rating,
sp.delivery_days

ORDER BY Revenue DESC;
"""

pd.read_sql(query, engine)
```

```python
#Top 10 Customers using RANK()
query = """
SELECT *

FROM (

SELECT

c.customer_name,

ROUND(SUM(s.revenue),2) AS Revenue,

RANK() OVER(
ORDER BY SUM(s.revenue) DESC
) AS Revenue_Rank

FROM sales_transactions s

JOIN customers c

ON s.customer_id = c.customer_id

GROUP BY c.customer_name

) Ranked_Customers

WHERE Revenue_Rank <=10;
"""

pd.read_sql(query, engine)
```

```python
#Top 10 Medicines using DENSE_RANK()
query = """
SELECT *

FROM (

SELECT

p.medicine_name,

ROUND(SUM(s.revenue),2) AS Revenue,

DENSE_RANK() OVER(
ORDER BY SUM(s.revenue) DESC
) AS Medicine_Rank

FROM sales_transactions s

JOIN products p

ON s.product_id=p.product_id

GROUP BY p.medicine_name

) Ranked_Medicines

WHERE Medicine_Rank<=10;
"""

pd.read_sql(query, engine)
```

```python
#Highest Revenue Order in Each City (ROW_NUMBER())
query = """
SELECT *

FROM (

SELECT

order_id,

city,

revenue,

ROW_NUMBER() OVER(

PARTITION BY city

ORDER BY revenue DESC

) AS Row_Num

FROM sales_transactions

) Ranked

WHERE Row_Num=1;
"""

pd.read_sql(query, engine)
```

```python
#Revenue above average
query = """
WITH AvgRevenue AS
(
SELECT
AVG(revenue) AS Avg_Revenue
FROM sales_transactions
)

SELECT

order_id,

customer_id,

revenue

FROM sales_transactions

WHERE revenue>

(
SELECT Avg_Revenue
FROM AvgRevenue
)

ORDER BY revenue DESC;
"""

pd.read_sql(query, engine)
```

```python
#Category revenue using CTE
query = """
WITH CategoryRevenue AS
(

SELECT

p.category,

SUM(s.revenue) AS Revenue

FROM sales_transactions s

JOIN products p

ON s.product_id=p.product_id

GROUP BY p.category

)

SELECT *

FROM CategoryRevenue

ORDER BY Revenue DESC;
"""

pd.read_sql(query, engine)
```

```python
#Customers spending above average
query = """
SELECT

customer_id,

ROUND(SUM(revenue),2) AS Revenue

FROM sales_transactions

GROUP BY customer_id

HAVING Revenue>

(

SELECT AVG(CustomerRevenue)

FROM

(

SELECT

SUM(revenue) AS CustomerRevenue

FROM sales_transactions

GROUP BY customer_id

) AvgTable

)

ORDER BY Revenue DESC;
"""

pd.read_sql(query, engine)
```

```python
#Product selling above average quantity
query = """
SELECT

product_id,

SUM(quantity) AS Quantity

FROM sales_transactions

GROUP BY product_id

HAVING Quantity>

(

SELECT AVG(ProductQty)

FROM

(

SELECT

SUM(quantity) AS ProductQty

FROM sales_transactions

GROUP BY product_id

) QtyTable

)

ORDER BY Quantity DESC;
"""

pd.read_sql(query, engine)
```

```python
#Running revenue
query = """
SELECT

order_date,

ROUND(SUM(revenue),2) AS Daily_Revenue,

ROUND(

SUM(SUM(revenue))

OVER(

ORDER BY order_date

)

,2)

AS Running_Revenue

FROM sales_transactions

GROUP BY order_date

ORDER BY order_date;
"""

pd.read_sql(query, engine)
```

```python
#revenue share by category
query = """
SELECT

p.category,

ROUND(SUM(s.revenue),2) AS Revenue,

ROUND(

SUM(s.revenue)*100/

(

SELECT SUM(revenue)

FROM sales_transactions

)

,2)

AS Revenue_Percentage

FROM sales_transactions s

JOIN products p

ON s.product_id=p.product_id

GROUP BY p.category

ORDER BY Revenue DESC;
"""

pd.read_sql(query, engine)
```
