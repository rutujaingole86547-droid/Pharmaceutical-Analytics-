```python
sales["month"] = (
sales["order_date"]
.dt.to_period("M")
)

sales.groupby(
"month"
)["revenue"].sum().plot(
figsize=(12,5)
)

plt.title(
"Monthly Revenue Trend"
)

plt.xlabel(
"Month"
)

plt.ylabel(
"Revenue"
)

plt.show()
```
